### MVF

$$
\min \frac{1}{2}\omega' \Sigma \omega   \;\;\;\;\;\;\;\;\; \text{s.t.      }  \omega' a = 1 , \text{    }a=[1,1,1, \cdots, 1]'
$$

\begin{aligned}
\mathcal{L}  &= \frac{1}{2}\omega' \Sigma \omega + \lambda \big( \omega' a -1 \big)\\
&\frac{\partial \mathcal{L}}{\partial \omega} = \Sigma \omega + \lambda a = 0  \;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;     (a)\\   
&\frac{\partial \mathcal{L}}{\partial \lambda} = \omega' a  - 1 = 0 \;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;\;   (b)\\
\end{aligned}


- from $(a)$: $\omega = \lambda \Sigma^{-1} a$
- and multiply both sides with $a'$, then $a'\omega = 1 = \lambda  a'\Sigma^{-1} a$
   - $\lambda = \frac{1}{ a'\Sigma^{-1} a}$

- substitute the above $\lambda$   into $(a)$, then $\omega = \frac{\Sigma^{-1}a}{ a'\Sigma^{-1} a}$


----

2. inverse variance portfolio
