## Eigenvalue and eigenvector

- for matrix A, $\lambda, x$ is eigenvalue and eigenvector, respectively. 
$$
\begin{align}
  &Ax = \lambda x \\
  &(A-\lambda I ) x = 0 \\
   &\rightarrow det(A-\lambda I) = 0
\end{align}
$$
- Example: 
    $
    A = 
     \begin{bmatrix}
        2 & 1\\
        1 & 2\\
     \end{bmatrix},  
    (A-\lambda I) =\begin{bmatrix}
        2-\lambda & 1\\
        1 & 2-\lambda\\
     \end{bmatrix}, det(A-\lambda I)=(2-\lambda)^2 -1 =0
    $
 
 $\Longrightarrow \lambda$ = 1, 3
 
$
(A-\lambda_1)
\begin{bmatrix}
x_1 \\ x_2
\end{bmatrix}
=\begin{bmatrix}
   2-1 & 1\\
   1 & 2-1\\
\end{bmatrix} 
\begin{bmatrix}
x_1 \\ x_2
\end{bmatrix},  
   (A-\lambda_2)
\begin{bmatrix}
x_1 \\ x_2
\end{bmatrix}
=\begin{bmatrix}
   2-3 & 1\\
   1 & 2-3\\
\end{bmatrix} 
\begin{bmatrix}
x_1 \\ x_2
\end{bmatrix}
$

     
$
    v_{\lambda_1} = 
    \begin{bmatrix}
    1 \\ -1
    \end{bmatrix}, 
     v_{\lambda_2} = 
    \begin{bmatrix}
    1 \\ 1
    \end{bmatrix}
 $

## Basic matrix terminology 

**Def: Unitary matrix**
- $U$ is unitary matrix if $U \cdot U^T=I$. That is $U^{-1}=U^T$

---
**Def: Singular matrix**
- $U$ is singular matrix if  $U$ is not convertible: $det(U)=0$
- example: A =[[3,6], [1,2]]

---
**Def: diagonal decomposition**

$\underset{n\times n}{A} = \underset{n\times n}{Q} \underset{n\times n}{D} \underset{n\times n}{Q} 
$

where 
- $Q=[q_1, q_2, \cdots, q_n]$ is A's eigenvectors
- $A$=(diagonal matrix) whose elements are eigenvalues of A 

---
**Def: Singular values**

- Let $A, A^*$ are matrix and its conjugate(for real it's transpose)
- $\sigma$ is called singular value and $u, v$ are corresponding singular vector if

$
\begin{aligned}
 &Av= \sigma u \\
 &A^* u = \sigma v
\end{aligned}$

- Singular values of the matrix A is the square root of the eigenvalues of the matrix (A multiplied by A transpose) or(A transpose multplied by A)

 - the two are identical with positive eigenvalues.
 - $A^* Av = \sigma A^* u = \sigma^2 v$
 - $A A^* v = \sigma^2 u$
   - Hence $\sigma^2$ is eigenvalue of $A^* A$ and $AA^* $

## Matrix decomposition: **LU decompostion**

- **A: square matrix**
    1. $A = LU$, where $L$=(low triangle matrix),  $U$=(upper triangle matrix)
    2. $A = LDU$, where $L$=(low triangle matrix),  $U$=(upper triangle matrix), $D$= (diagonal matrix)
    3. $A = LUP$, where $L$=(low triangle matrix),  $U$=(upper triangle matrix), $P$= (permutation matrix)

---
## Rank factorization

- **$A_{m \times n}$** with rank $r$
  1. $A=CF$, where $C=m \times r$ with full column rank $r$, $F= r \times n$ with full row rank $r$

---
## Cholesky decompostion

- **$A$ is square or postive semi-definite matrix**
  1. $A=U^* U$ where $U$ is upper triangular with real positive diagonal, $U^*$ is cojugate matrix of $U$

---
## QR decompostion
-  **$A$ is $m \times n$ matrix **
   1. $A=QR$ where $Q$ is $(m \times m)$ unitary matrix, $R$ is $(m \times n)$ upper triangular matrix

---  
## Eigendecompoistion (= spectral decomposition)

- **$A$ is square matrix **
   1. $A = V D V^{-1}$ where  columns of $V$ are eigenvectors of $A$, diagonal marix D is formed with eigenvalues of $A$


--- 
## Singular value decomposition

- **$A$ is $(m \times n)$ matrix **  
  1. $A = U D V^*$ where $D$ is non-negative diagonal matrix, $U$ and $V$ are unitary matrices, $V^*$ is the conjugate of V
  2. The diagonal elements of $D$ are called singular values of $A$
