# Standard Procedure

-  Standardized on the original data
    - either demean or decompose correlation/covariance matrix
-  Create correlation matrix of the standardized data
-  Find eigen vectors and values  of the correlation matrix
-  Determine the number of main principal component
-  Applying evaluation function

-  standardized : PCA is very sensitive to normalization or not
    $$
      z_i = \frac{x_i -\bar{x}}{\sigma}
    $$

- correlation matrix
    $$
        \Sigma_z =
            \begin{bmatrix}
                r_{11} & r_{12} & \cdots & r_{1n}\\
                r_{21} & r_{22} & \cdots & r_{2n}\\
                & \vdots & \\
                r_{n1} & r_{n2} & \cdots & r_{nn}
            \end{bmatrix}
    $$

- Eigenvalue($\lambda's$) and Eigenvectors($Q$)

 $$        \Sigma_z = QRQ =
            Q\begin{bmatrix}
                \lambda_1 &  &  & \\
                          & \lambda_2 &  &\\
                &  & \ddots & \\
                &  &  & \lambda_p
            \end{bmatrix} Q
$$

-  Each Component's variance contribution

$$
     \frac{\lambda_k}{\sum_{i=1}^{p}\lambda_i }
$$


-  In PCA should decide number of eigenvalue or explained variance
$$
    v_k = \frac{\sum_{i=1}^{k}\lambda_i }{\sum_{i=1}^{p}\lambda_i }  \ge 90\%
$$

Principal Component Analysis (PCA) is a statistical method used to reduce dimensionality and to identify patterns in data. In the context of financial markets, PCA can be applied to various types of data such as stock prices, interest rates, or other financial metrics. Here's a simplified outline on how PCA might be applied to financial market data:

1. **Data Collection and Preparation**:
    - Collect historical data: Gather historical financial data such as stock prices, trading volumes, or other relevant financial metrics over a certain period.
    - Normalize the data: Scale the data so that each financial metric has a mean of 0 and a variance of 1.

2. **Performing PCA**:
    - Calculate the covariance matrix: Create a covariance matrix of the normalized data to understand the relationships between different financial metrics.
    - Calculate eigenvectors and eigenvalues: Compute the eigenvectors and eigenvalues of the covariance matrix. The eigenvectors represent the directions of maximum variance, and the eigenvalues indicate the magnitude of the variance in each direction.

3. **Analysis of Principal Components**:
    - Select principal components: Choose the number of principal components to retain based on the cumulative explained variance. Typically, you would select the first few principal components that explain a significant portion of the variance in the data.
    - Interpret principal components: Each principal component is a linear combination of the original financial metrics. Interpret the coefficients to understand what each principal component represents in terms of the original financial metrics.

4. **Utilization of Principal Components**:
    - Risk Assessment: Use the principal components to assess the systematic risk in a portfolio or to understand the common factors driving the variances in different financial instruments.
    - Portfolio Optimization: Utilize the principal components to construct portfolios that optimize for certain characteristics, such as minimizing variance or maximizing return.
    - Asset Pricing Models: Incorporate the principal components into asset pricing models to better explain the returns of different financial assets.

5. **Visualization and Communication**:
    - Create visualizations: Use scatter plots, bar charts, or other visualizations to communicate the findings from the PCA analysis.
    - Report findings: Prepare reports or presentations to communicate the insights gained from the PCA analysis to stakeholders.

6. **Continuous Monitoring and Updating**:
    - Monitor changes: Continuously monitor the financial market for significant changes and update the PCA analysis accordingly.
    - Update stakeholders: Keep stakeholders updated with any significant findings from the updated PCA analysis.

PCA can be a complex process and requires a solid understanding of statistics and financial concepts. Moreover, it's essential to validate the assumptions and results of PCA, and to consider consulting with financial analysts or statisticians if necessary.
